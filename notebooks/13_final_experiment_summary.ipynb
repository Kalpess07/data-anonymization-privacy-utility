```python

from pathlib import Path
import pandas as pd

cwd = Path.cwd()
PROJECT_ROOT = cwd.parent if cwd.name.lower() == "notebooks" else cwd
RESULTS_DIR = PROJECT_ROOT / "results"

# Load baseline and Level 1 experiment results
baseline = pd.read_csv(RESULTS_DIR / "baseline_metrics.csv")
lr_level1 = pd.read_csv(
    RESULTS_DIR / "day10_level1_logistic_regression.csv"
)
rf_level1 = pd.read_csv(
    RESULTS_DIR / "day11_level1_random_forest.csv"
)

# Standardize result columns
baseline_summary = baseline[
    ["model", "accuracy", "precision", "recall", "f1_score"]
].copy()
baseline_summary["experiment"] = "Baseline"
baseline_summary["anonymization_level"] = "none"

lr_summary = lr_level1[
    ["model", "accuracy", "precision", "recall", "f1_score"]
].copy()
lr_summary["experiment"] = "Level 1"
lr_summary["anonymization_level"] = "Level 1"

rf_summary = rf_level1[
    ["model", "accuracy", "precision", "recall", "f1_score"]
].copy()
rf_summary["experiment"] = "Level 1"
rf_summary["anonymization_level"] = "Level 1"

# Combine all four experiments
final_summary = pd.concat(
    [baseline_summary, lr_summary, rf_summary],
    ignore_index=True
)

final_summary = final_summary[
    [
        "experiment",
        "anonymization_level",
        "model",
        "accuracy",
        "precision",
        "recall",
        "f1_score"
    ]
]

output_path = RESULTS_DIR / "final_experiment_summary.csv"
final_summary.to_csv(output_path, index=False)

print("Saved:", output_path)
display(final_summary.round(4))
```
