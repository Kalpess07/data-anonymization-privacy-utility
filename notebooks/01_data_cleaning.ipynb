```python
#Step 1 — Import libraries and set up paths

import pandas as pd
import numpy as np
from pathlib import Path
```

```python

# Detect the project root
cwd = Path.cwd()
```

```python
if cwd.name.lower() == "notebooks":
    PROJECT_ROOT = cwd.parent
else:
    PROJECT_ROOT = cwd
```

```python
# Define paths
RAW_PATH = PROJECT_ROOT / "data" / "raw.csv"
CLEAN_DIR = PROJECT_ROOT / "data" / "cleaned"
```

```python
# Create output directory
CLEAN_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Verify paths
print("Project root:", PROJECT_ROOT)
print("Raw dataset exists:", RAW_PATH.exists())
print("Cleaned directory:", CLEAN_DIR)
```

```python
#Step 2 — Inspect the raw CSV before loading it

# Display the first five lines of the CSV
with open(RAW_PATH, "r", encoding="utf-8-sig") as file:
    for _ in range(5):
        line = file.readline()
        if not line:
            break
        print(line.rstrip())
```

```python
# Inspect the number of columns without assuming headers
preview = pd.read_csv(
    RAW_PATH,
    header=None,
    nrows=5,
    skipinitialspace=True
)

print("Number of columns:", preview.shape[1])
display(preview.head())
```

```python
#Step 3.1 — Load the dataset correctly

adult_columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income",
    "zip_code"
]
```

```python
#Step 3.2 — Load the dataset correctly

df = pd.read_csv(
    RAW_PATH,
    header=0,
    skipinitialspace=True,
    na_values=["?"],
    dtype={"zip_code": "string"}
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
#Step 4 — Inspect the dataset

print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("Data Type"))

print("\nMissing values per column:")
display(df.isna().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nIncome distribution:")
display(df["income"].value_counts(dropna=False).to_frame("Count"))
```

```python
#Step 5 — Standardize column names and clean text values

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace("-", "_", regex=False)
    .str.replace(" ", "_", regex=False)
)

# Remove unnecessary whitespace from text columns
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df[col] = df[col].str.strip()

# Convert remaining '?' markers into missing values
df = df.replace(r"^\s*\?\s*$", pd.NA, regex=True)

print("Column names standardized.")
print("Missing values after normalization:")
display(df.isna().sum().to_frame("Missing Values"))
```

```python
#Step 6 — Remove duplicate rows

rows_before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

rows_removed = rows_before - len(df)

print("Rows before:", rows_before)
print("Duplicate rows removed:", rows_removed)
print("Rows after:", len(df))
```

```python
# Step 7 — Validate and save the cleaned dataset

# Validate expected columns
expected_columns = {
    "age", "workclass", "income", "zip_code"
}

assert expected_columns.issubset(df.columns)
assert not df.duplicated().any()

print("Validation passed.")
print("Final shape:", df.shape)
print("Target column:", "income")

# Save cleaned data
CLEAN_PATH = CLEAN_DIR / "adult_cleaned.csv"

df.to_csv(CLEAN_PATH, index=False)

print("Saved to:", CLEAN_PATH)
```
